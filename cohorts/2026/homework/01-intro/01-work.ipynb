```python
# 1. 原始中位数
median_before = df['horsepower'].median()

# 2. 最常见值，也就是众数
mode_hp = df['horsepower'].mode()[0]

# 3. 用众数填充缺失值
df['horsepower'] = df['horsepower'].fillna(mode_hp)

# 4. 再次计算中位数
median_after = df['horsepower'].median()

print('median_before =', median_before)
print('mode_hp =', mode_hp)
print('median_after =', median_after)

if median_after > median_before:
    print("Yes, it increased")
elif median_after < median_before:
    print("Yes, it decreased")
else:
    print("No")
```

```python
# 1. 选亚洲车
asia = df[df['origin'] == 'Asia']

# 2. 选两列
X_df = asia[['vehicle_weight', 'model_year']]

# 3. 取前 7 行
X_df = X_df.head(7)

# 4. 转成 NumPy 数组
X = X_df.to_numpy()

# 5. 计算 X.T @ X
XTX = X.T @ X

# 6. 求逆
XTX_inv = np.linalg.inv(XTX)

# 7. 创建 y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. 计算 w
w = XTX_inv @ X.T @ y

# 9. 求和
print('w =', w)
print('sum =', w.sum())
```
